# Результаты этапа 1
Просмотр сохранённых метрик и выборки изображений. Для полной генерации см. `src/` и `configs/baseline.json`.

```python
import json
from pathlib import Path
path = Path('results/summary.json')
summary = json.loads(path.read_text())
for split in ('dev','test','all'):
    s = summary['summaries'][split]
    print(split, 'images=',s['generated_images'],
          'Edge F1=',round(s['edge_f1']['mean'],4),
          'DINO=',round(s['dino_cosine']['mean'],4),
          'time(s)=',round(s['seconds']['mean'],2))
```
